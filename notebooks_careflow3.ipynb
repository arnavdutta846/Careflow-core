```python
mamba install pandas
import pandas as pd
df = pd.read_csv('healthcare.csv')
df
```

```python
#cheking any null values present in the data
df.isnull().sum()
```

```python
#Cleaning the data for null values
df=df.dropna(subset=['Department Referral'])
df=df.dropna(subset=['Patient Satisfaction Score'])
```

```python
df.isnull().sum()
```

```python
#Cheaking 10 rows 
df.head(10)
```

```python
#cheaking what are the columns present
df.columns
```

```python
#Reducing the average waittime to 20 min
df["Improved Waittime"] = df["Patient Waittime"] - 20
df["Improved Waittime"] = df["Improved Waittime"].clip(lower=10)

print(df["Improved Waittime"].mean())
```

```python
df["Admission_Time"] = pd.to_datetime(
    df["Patient_Admission_Time"],
    format="%I:%M:%S %p")
```

```python
df["Hour"] = df["Admission_Time"].dt.hour
```

```python
df.columns=df.columns.str.replace(' ','_')
df
```

```python
admissions_by_hour = df.groupby("Hour").size()
print(admissions_by_hour)
```

```python
peak_hour = admissions_by_hour.idxmax()
peak_count = admissions_by_hour.max()

print("Peak admission hour:", peak_hour)
print("Number of admissions:", peak_count)
```

```python
max_age = df["Patient_Age"].max()

print("Maximum age:", max_age, "years")
```
