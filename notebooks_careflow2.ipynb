```python
import pandas as pd
df=pd.read_csv('healthcare.csv')
df
```

```python
df.isnull().sum()
```

```python
df = df .dropna(subset=['Department Referral'])
```

```python
df = df .dropna(subset=['Patient Satisfaction Score'])
```

```python
df.isnull().sum()
```

```python
df.head(10)
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
df
```

```python
df.columns
```

```python
#Reducing average wait time by 20 mins
df["improved_waittime"] = df["patient_waittime"] - 20
df["improved_waittime"] = df["improved_waittime"].clip(lower=10)

print(df["improved_waittime"].mean())
```

```python
df.head(10)
```

```python
#top admission hour
df["admission_time"] = pd.to_datetime(
    df["patient_admission_time"],
    format="%I:%M:%S %p")
```

```python
df["hour"] = df["admission_time"].dt.hour
```

```python
admissions_by_hour = df.groupby("hour").size()
print(admissions_by_hour)
```
